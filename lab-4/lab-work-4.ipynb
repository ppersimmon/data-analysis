```python
from pymongo import MongoClient
```

```python
client = MongoClient('mongodb://localhost:27017/')
client.list_database_names()
```

```python
mydb = client['mydb']
mydb.list_collection_names()
```

```python
products = mydb['products']
```

```python
for product in products.find().limit(5):
    print(product)
```

```python
for product in products.find({"category": "Clothing"}):
    print(product)
```

```python
for product in products.find({"price": {"$lt" : 15}, "product_name": "Eggs"}):
    print(product)
```

```python
for product in products.find({"product_name": "Eggs"}):
    print(product)
```

```python
for product in products.find().sort("description", 1).limit(15):
    print(product)
```

```python
for product in products.find().sort("arrival_date", -1).limit(15):
    print(product)
```

```python
import datetime
products.insert_one({"_id": 1001, 'product_name': 'Soap', 'category': 'Health & Beauty', 'seller': 'Seller2', 'price': 40.71, 'arrival_date': datetime.datetime(2026, 3, 21, 0, 0)})
```
